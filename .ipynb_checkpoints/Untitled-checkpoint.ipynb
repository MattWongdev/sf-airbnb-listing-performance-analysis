# San Francisco Airbnb Listing Performance Analysis

## Business Question: Which listing and host features are associated with higher occupancy for Airbnb listings in San Francisco?

### Project Context

This project focuses on listing performance rather than simple descriptive EDA. The goal is to identify patterns in listing structure, host behavior, and location that are associated with stronger occupancy outcomes in San Francisco.

### Data Notes

The dataset includes listing details, review metrics, host signals, and Inside Airbnb occupancy estimates. In this extract, `price` and `estimated_revenue_l365d` are missing, so the analysis is centered on occupancy instead of pricing or revenue.

### Cleaning Strategy

The cleaning approach is intentionally light. Binary host flags are converted to numeric values, percentage fields are converted to numeric form, and missing values are retained unless a specific analysis requires that variable. This preserves as much of the dataset as possible while keeping the transformations easy to explain.

```python
#Importing Libraries and csv file
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("sf_listings.csv")
```

```python
df.info()
```

```python
#Picking the columns most useful for analysis
cols = [
    'id',
    'name',
    'room_type',
    'property_type',
    'neighbourhood_cleansed',
    'accommodates',
    'bedrooms',
    'beds',
    'minimum_nights',
    'number_of_reviews',
    'reviews_per_month',
    'review_scores_rating',
    'host_is_superhost',
    'instant_bookable',
    'host_response_rate',
    'host_acceptance_rate',
    'calculated_host_listings_count',
    'availability_365',
    'estimated_occupancy_l365d'
]

airbnb = df[cols].copy()
airbnb.head()
```

```python
airbnb.info()
```

```python
airbnb[['host_is_superhost', 'instant_bookable']] = (
    airbnb[['host_is_superhost', 'instant_bookable']]
    .replace({'t': 1, 'f': 0})
)

airbnb[['host_response_rate', 'host_acceptance_rate']] = (
    airbnb[['host_response_rate', 'host_acceptance_rate']]
    .replace('%', '', regex=True)
    .apply(pd.to_numeric, errors='coerce')
)

```

```python
airbnb.isnull().sum().sort_values(ascending=False)
```

```python
airbnb['occupied_nights'] = airbnb['estimated_occupancy_l365d']
airbnb['occupancy_rate'] = airbnb['occupied_nights'] / 365

airbnb[['occupied_nights', 'occupancy_rate']].describe()
```

```python
airbnb.info()
airbnb.isna().sum().sort_values(ascending=False)
```

### Exploratory Analysis

```python
airbnb['occupied_nights'].hist(bins=40)
plt.title("Distribution of Estimated Occupied Nights")
plt.xlabel("Occupied Nights in Last 365 Days")
plt.ylabel("Number of Listings")
plt.show()
```

The distribution appears right-skewed, with many listings clustered at relatively low occupancy and a smaller group achieving much stronger booking performance. This suggests occupancy is uneven across the San Francisco market, which makes feature-level comparisons more useful than relying on a single overall average.

```python
room_occ = (
    airbnb.groupby('room_type')['occupancy_rate']
    .median()
    .sort_values()
)

room_occ.plot(kind='barh')
plt.title("Median Occupancy Rate by Room Type")
plt.xlabel("Median Occupancy Rate")
plt.ylabel("Room Type")
plt.show()

room_occ
```

Room type creates a clear occupancy gap in this dataset. Shared rooms and entire homes show higher median occupancy rates than private rooms, although the shared-room category should be interpreted carefully because it contains far fewer listings. Entire homes appear to combine stronger occupancy with a much larger sample size, making that pattern more credible.

```python

# ADSFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFFF
neigh_occ = (
    airbnb.groupby('neighbourhood_cleansed')
    .agg(
        listings=('id', 'count'),
        median_occupancy_rate=('occupancy_rate', 'median')
    )
    .query('listings >= 15')
    .sort_values('median_occupancy_rate', ascending=False)
    .head(15)
)

neigh_occ
```

```python
neigh_occ.sort_values('median_occupancy_rate').plot(
    kind='barh',
    y='median_occupancy_rate',
    legend=False
)
plt.title("Top Neighborhoods by Median Occupancy Rate")
plt.xlabel("Median Occupancy Rate")
plt.ylabel("Neighborhood")
plt.show()
```

Neighborhood appears to matter meaningfully, with several areas standing out even after filtering for a minimum number of listings. Using median occupancy rate instead of raw occupied nights makes these comparisons easier to interpret and better aligned with the project goal.

```python
airbnb['capacity_group'] = pd.cut(
    airbnb['accommodates'],
    bins=[0, 2, 4, 6, 16],
    labels=['1-2 guests', '3-4 guests', '5-6 guests', '7+ guests']
)

acc_occ = (
    airbnb.groupby('capacity_group', observed=False)['occupancy_rate']
    .median()
    .dropna()
)

acc_occ.plot(kind='bar')
plt.title("Median Occupancy Rate by Guest Capacity Group")
plt.xlabel("Capacity Group")
plt.ylabel("Median Occupancy Rate")
plt.show()

acc_occ
```

Grouping capacity makes this comparison easier to trust than plotting every exact `accommodates` value. The binned view reduces noise from rare listing sizes and gives a clearer read on whether small, medium, or large listings tend to maintain stronger occupancy.

```python
#REVIEW THIS

summary = (
    airbnb.groupby('room_type')
    .agg(
        listings=('id', 'count'),
        median_occ_nights=('occupied_nights', 'median'),
        median_occ_rate=('occupancy_rate', 'median'),
        median_reviews=('number_of_reviews', 'median'),
        median_rating=('review_scores_rating', 'median')
    )
    .sort_values('median_occ_rate', ascending=False)
)

summary
```

### Additional Occupancy Drivers

```python
airbnb.boxplot(column='occupancy_rate', by='host_is_superhost', grid=False)
plt.title("Occupancy Rate by Superhost Status")
plt.suptitle("")
plt.xlabel("Superhost (0 = No, 1 = Yes)")
plt.ylabel("Occupancy Rate")
plt.show()

airbnb.groupby('host_is_superhost')['occupancy_rate'].median()
```

Superhost status is one of the most practical host-side signals to test because it reflects responsiveness, consistency, and guest experience. A visible separation here would support the idea that stronger host quality is associated with better booking performance.

```python
airbnb.boxplot(column='occupancy_rate', by='instant_bookable', grid=False)
plt.title("Occupancy Rate by Instant Bookable Status")
plt.suptitle("")
plt.xlabel("Instant Bookable (0 = No, 1 = Yes)")
plt.ylabel("Occupancy Rate")
plt.show()

airbnb.groupby('instant_bookable')['occupancy_rate'].median()


airbnb['min_stay_group'] = pd.cut(
    airbnb['minimum_nights'],
    bins=[0, 1, 2, 6, 29, 3650],
    labels=['1 night', '2 nights', '3-6 nights', '7-29 nights', '30+ nights']
)

min_occ = (
    airbnb.groupby('min_stay_group', observed=False)['occupancy_rate']
    .median()
    .dropna()
)

min_occ.plot(kind='bar', color='steelblue')
plt.title("Median Occupancy Rate by Minimum Stay Requirement")
plt.xlabel("Minimum Stay Group")
plt.ylabel("Median Occupancy Rate")
plt.xticks(rotation=30)
plt.show()

min_occ
```

```python
room_min_occ = (
    airbnb.groupby(['room_type', 'min_stay_group'])['occupancy_rate']
    .median()
    .unstack()
)

room_min_occ.T.plot(kind='bar', figsize=(10, 6))
plt.title("Median Occupancy Rate by Minimum Stay Group and Room Type")
plt.xlabel("Minimum Stay Group")
plt.ylabel("Median Occupancy Rate")
plt.xticks(rotation=30)
plt.legend(title="Room Type")
plt.show()

room_min_occ
```

```python
top_neighborhoods = (
    airbnb['neighbourhood_cleansed']
    .value_counts()
    .head(10)
    .index
)

superhost_neigh = (
    airbnb[airbnb['neighbourhood_cleansed'].isin(top_neighborhoods)]
    .groupby(['neighbourhood_cleansed', 'host_is_superhost'])['occupancy_rate']
    .median()
    .unstack()
)

superhost_neigh.plot(kind='bar', figsize=(12, 6))
plt.title("Median Occupancy Rate by Neighborhood and Superhost Status")
plt.xlabel("Neighborhood")
plt.ylabel("Median Occupancy Rate")
plt.xticks(rotation=45, ha='right')
plt.legend(title="Superhost", labels=['No', 'Yes'])
plt.tight_layout()
plt.show()

superhost_neigh
```

```python
accept_df = airbnb.dropna(subset=['host_acceptance_rate', 'occupancy_rate']).copy()

accept_df['acceptance_rate_group'] = pd.cut(
    accept_df['host_acceptance_rate'],
    bins=[0, 50, 75, 90, 100],
    labels=['0-50', '51-75', '76-90', '91-100']
)

superhost_accept = (
    accept_df.groupby(['acceptance_rate_group', 'host_is_superhost'], observed=False)['occupancy_rate']
    .median()
    .unstack()
)

superhost_accept.plot(kind='bar', figsize=(10, 6))
plt.title("Median Occupancy Rate by Acceptance Rate Group and Superhost Status")
plt.xlabel("Host Acceptance Rate Group")
plt.ylabel("Median Occupancy Rate")
plt.xticks(rotation=0)
plt.legend(title="Superhost", labels=['No', 'Yes'])
plt.show()

superhost_accept
```

```python
rpm_df = airbnb.dropna(subset=['reviews_per_month', 'occupancy_rate']).copy()

rpm_df['reviews_pm_group'] = pd.cut(
    rpm_df['reviews_per_month'],
    bins=[0, 0.5, 1, 2, 5, 50],
    labels=['0-0.5', '0.5-1', '1-2', '2-5', '5+']
)

superhost_rpm = (
    rpm_df.groupby(['reviews_pm_group', 'host_is_superhost'], observed=False)['occupancy_rate']
    .median()
    .unstack()
)

superhost_rpm.plot(kind='bar', figsize=(10, 6))
plt.title("Median Occupancy Rate by Reviews per Month Group and Superhost Status")
plt.xlabel("Reviews per Month Group")
plt.ylabel("Median Occupancy Rate")
plt.xticks(rotation=0)
plt.legend(title="Superhost", labels=['No', 'Yes'])
plt.show()

superhost_rpm
```

```python
superhost_host_quality = (
    airbnb.groupby('host_is_superhost')
    .agg(
        listings=('id', 'count'),
        median_occupancy_rate=('occupancy_rate', 'median'),
        median_response_rate=('host_response_rate', 'median'),
        median_acceptance_rate=('host_acceptance_rate', 'median'),
        median_reviews=('number_of_reviews', 'median'),
        median_reviews_per_month=('reviews_per_month', 'median')
    )
)

superhost_host_quality
```

The instant-bookable comparison suggests that booking flexibility alone does not explain stronger occupancy in this dataset. Minimum-stay rules show an interesting overall pattern, but the deeper subgroup checks below suggest that the apparent advantage of a 2-night minimum is not fully consistent across listing types.

Breaking minimum-stay groups out by room type provides a better test of whether the 2-night result is robust. The pattern weakens here: 2 nights remains strong in some categories, but it is not universally dominant, which suggests the original result was partly shaped by listing composition.

The neighborhood comparison makes the superhost pattern much more convincing. In most high-volume neighborhoods, superhost listings have meaningfully higher median occupancy than non-superhost listings, which suggests the effect is not limited to just one part of the city.

The acceptance-rate bins reinforce the superhost story rather than replacing it. Occupancy tends to improve as host acceptance rises, and superhosts remain above non-superhosts within every acceptance-rate group, making superhost status look like a consistently positive signal.

Reviews-per-month is strongly associated with occupancy, which is expected because listings with more booking activity also tend to generate more reviews. The superhost gap is still visible in lower and mid review-activity groups, but the relationship should be interpreted as a strong association rather than clear evidence of causation.

```python
superhost_detail = (
    airbnb[airbnb['neighbourhood_cleansed'].isin(top_neighborhoods)]
    .groupby(['neighbourhood_cleansed', 'room_type', 'host_is_superhost'])
    .agg(
        listings=('id', 'count'),
        median_occupancy_rate=('occupancy_rate', 'median')
    )
    .query('listings >= 10')
    .sort_values('median_occupancy_rate', ascending=False)
)

superhost_detail.head(20)
```

The host-quality summary table is useful for combining several signals in one place. Taken together, it supports the same direction seen in the charts: superhosts tend to have stronger occupancy, higher acceptance and response behavior, and stronger review activity than non-superhosts.

The detailed neighborhood-room-type table is best treated as supporting evidence rather than a headline result. It shows that many of the strongest occupancy combinations come from superhost listings, which is consistent with the broader charts, but the table is mainly helpful for spotting specific high-performing segments.

Review quality and correlation checks help round out the analysis, but they should be interpreted carefully. Review scores appear only weakly related to occupancy on their own, while the heatmap suggests that review activity, host acceptance behavior, and superhost-related signals are more strongly associated with stronger booking performance.

```python
review_df = airbnb.dropna(subset=['review_scores_rating', 'occupancy_rate']).copy()

plt.figure(figsize=(8, 5))
plt.scatter(review_df['review_scores_rating'], review_df['occupancy_rate'], alpha=0.25)
plt.title("Review Rating vs Occupancy Rate")
plt.xlabel("Review Score Rating")
plt.ylabel("Occupancy Rate")
plt.show()


corr_cols = [
    'occupancy_rate',
    'accommodates',
    'minimum_nights',
    'number_of_reviews',
    'reviews_per_month',
    'review_scores_rating',
    'host_response_rate',
    'host_acceptance_rate',
    'calculated_host_listings_count',
    'availability_365'
]

corr = airbnb[corr_cols].corr()

plt.figure(figsize=(10, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Correlation Heatmap of Occupancy and Numeric Features")
plt.show()
```
